````python
%%sql

CREATE TABLE employees (
    id SERIAL,
    email VARCHAR(255) NOT NULL,
    full_name VARCHAR(255) NOT NULL,
    salary NUMERIC(10, 2) NOT NULL DEFAULT 0,
    hired_date DATE NOT NULL DEFAULT CURRENT_DATE,
    left_on DATE,
    dept_id INT,   -- Foreign key
    manager_id BIGINT,
    status CHAR(8) NOT NULL DEFAULT 'active',
    CONSTRAINT pk_employees PRIMARY KEY(id),
    CONSTRAINT uq_employee_email UNIQUE (email),
    CONSTRAINT chk_salary CHECK (salary >= 0),
    CONSTRAINT check_date CHECK (left_on IS NULL OR left_on >= hired_date),
    CONSTRAINT fk_emp_dept FOREIGN KEY (dept_id) REFERENCES departments (id),
    CONSTRAINT fk_emp_manager FOREIGN KEY (manager_id)
        REFERENCES employees (id)
);
````
